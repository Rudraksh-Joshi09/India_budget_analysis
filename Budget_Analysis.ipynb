```python
import os
import re
import pdfplumber
import pandas as pd 
import numpy as np
```

```python
os.chdir('..')

os.chdir('Data Files')

# Now file is in current directory
pdfpath = "/Users/apple/Desktop/Data Science/Data Files"

with pdfplumber.open("Union Budget_2021-22_0.pdf") as pdf:
    page = pdf.pages[2]   # page 3
    text = page.extract_text()


lines = text.split("\n")
lines = lines[20:43]
lines
```

```python
data = []

for line in lines:
    
    nums = re.findall(r'-?[\d,.]+%?', line)
    
    # keep only rows with at least 4 numbers
    if len(nums) >= 4:
        
        # keep only first 5 numbers
        nums = nums[:5]
        
        category = re.sub(r'-?[\d,.]+%?', '', line).strip()
        
        data.append([category] + nums)

budget_21_22 = pd.DataFrame(data, columns=[
    "Category",
    "Actuals_2019-20",
    "Budgeted_2020-21",
    "Revised_2020-21",
    "Budgeted_2021-22",
    "Change"
])
budget_21_22["year"] = "2021-22"
budget_21_22 = budget_21_22.replace({",": "", "%": ""}, regex=True)

budget_21_22 =budget_21_22.drop(index=0)
budget_21_22.iloc[11,0]= "Revenue Deficit(% of GDP)"
budget_21_22.iloc[13,0]= "Fiscal Deficit(% of GDP)"
budget_21_22.iloc[15,0]= "Primary Deficit(% of GDP)"



budget_21_22
```

```python
def extract_budget_data(lines):
    data = []

    for line in lines:
        num = re.findall(r'-?[\d,.]+%?', line)

        if len(num) >= 4:
            num = num[:5]

            category = re.sub(r'-?[\d,.]+%?', "", line).strip()

            data.append([category] + num)

    df = pd.DataFrame(data, columns=[
        "Category",
        "Actual_2020-21",
        "Budgeted_2021-22",
        "Revised_2021-22",
        "Budgeted_2022-23",
        "Change_From_Previous_Year"
    ])
   
    df = df.replace({",": "", "%": ""}, regex=True)
   
    return df

budget_22_23 = extract_budget_data(lines)
budget_22_23["year"] = "2022-23"


budget_22_23.iloc[12,0]= "Revenue Deficit(% of GDP)"
budget_22_23.iloc[14,0]= "Fiscal Deficit(% of GDP)"
budget_22_23.iloc[16,0]= "Primary Deficit(% of GDP)"


budget_22_23 =budget_22_23.drop(index=0)

budget_22_23
```

```python
with pdfplumber.open(os.path.join(pdfpath, "Union_Budget_2026-27.pdf")) as pdf:

    page = pdf.pages[2]
    text = page.extract_text()

lines = text.split("\n")
lines = lines[19:38]
lines

budget_26_27 = extract_budget_data(lines)
budget_26_27.rename(columns={
    "Actuals_2020-21":"Actual_2024-25",
     "Budgeted_2021-22":"Budgeted_2025-26",
     "Revised_2021-22":"Revised_2025-26",
     "Budgeted_2022-23":"Budgeted_2026-27",
    "Change":"Change_From_Previous_Year"}, inplace=True)

budget_26_27["year"] = "2026-27"

budget_26_27.iloc[12,0]= "Revenue Deficit(% of GDP)"
budget_26_27.iloc[14,0]= "Fiscal Deficit(% of GDP)"
budget_26_27.iloc[16,0]= "Primary Deficit(% of GDP)"


budget_26_27 =budget_26_27.drop(index=0)

budget_26_27
```

```python
def to_df(obj):
    if isinstance(obj, pd.io.formats.style.Styler):
        return obj.data
    return obj

# Convert all 6
budget_21_22 = to_df(budget_21_22)
budget_22_23 = to_df(budget_22_23)
budget_23_24 = to_df(budget_23_24)
budget_24_25 = to_df(budget_24_25)
budget_25_26 = to_df(budget_25_26)
budget_26_27 = to_df(budget_26_27)

budget_21_22.columns = ["Category","Actual","Budgeted","Revised","Next_Budget","Change","Year"]

budget_22_23.columns = ["Category","Actual","Budgeted","Revised","Next_Budget","Change","Year"]

budget_23_24.columns = ["Category","Actual","Budgeted","Revised","Next_Budget","Change","Year"]

budget_24_25.columns = ["Category","Actual","Budgeted","Revised","Next_Budget","Change","Year"]

budget_25_26.columns = ["Category","Actual","Budgeted","Revised","Next_Budget","Change","Year"]

budget_26_27.columns = ["Category","Actual","Budgeted","Revised","Next_Budget","Change","Year"]



merged_budget = pd.concat(
    [budget_21_22, budget_22_23, budget_23_24,
     budget_24_25, budget_25_26, budget_26_27],
    ignore_index=True
)

# For Diplaying it in the aligned Center manner

merged_budget.style.set_properties(**{'text-align': 'center'}) \
    .set_table_styles([{
        'selector': 'th',
        'props': [('text-align', 'center')]
    }])

# merged_budget.iloc[:55]

# print("Shape:", merged_budget.shape)
# # Should be (96, 7) as 96 rows and 7 columns

# print("\nRows per year:")
# print(merged_budget['Year'].value_counts().sort_index())
# # Should show 16 for each year

# print("\nFirst few rows:")
# print(merged_budget.head(5).to_string())
```

```python
# DATA CLEANING 


merged_budget['Category'] = merged_budget['Category'].replace({
    'Capital outlay'                              : 'Capital Outlay',
    'Loans'                                       : 'Loans and Advances',
    'Loans and Advances and Advances'             : 'Loans and Advances',
    'Recoveries of Loans'                         : 'Recoveries of Loans and Advances',
    'Recoveries of Loans*'                        : 'Recoveries of Loans and Advances',
    'Other receipts (including disinvestments)*'  : 'Other receipts (including disinvestments)',
    'Disinvestments'                              : 'Other receipts (including disinvestments)',
    'Total Receipts (without borrowings)'         : 'Total Receipts (excluding borrowings)'
})

merged_budget['metric_type'] = merged_budget['Category'].apply(
    lambda x: 'Gdp_ratio' if '% of GDP' in x else 'Amount'
)


merged_budget.replace('N/A', np.nan, inplace=True)
merged_budget.replace('nan', np.nan, inplace=True)

#merged_budget.dtypes

num_cols = ['Actual', 'Budgeted', 'Revised', 'Next_Budget', 'Change']
for col in num_cols:
    merged_budget[col] = pd.to_numeric(merged_budget[col], errors='coerce')

merged_budget.iloc[:55]

```

```python
def classify(category):
    if '% of GDP' in category:
        return 'Deficit'
    elif 'Receipts' in category or 'Recoveries' in category or 'Disinvestment' in category or 'disinvestment' in category or 'Other receipts' in category:
        return 'Receipts'
    elif 'Expenditure' in category or 'Outlay' in category or 'Loans and Advances' in category:
        return 'Expenditure'
    elif 'Deficit' in category:
        return 'Deficit'
    else:
        return 'Other'

merged_budget['group'] = merged_budget['Category'].apply(classify)

merged_budget = merged_budget.reset_index(drop=True)
```
